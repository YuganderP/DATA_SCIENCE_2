```python
import numpy as np
import pandas as pd

df=pd.DataFrame({
    'age':np.random.randint(18,70,size=10).astype(float),
    'salary':np.random.normal(5000,15000,size=10),
    'city':["A","B","C","D","A","C","B","C","A","D"],
    'join_data':pd.date_range("2020-01-01",periods=10)
})
```

```python
df.loc[[1,2,5],'age']=np.nan
df.loc[[2,7],'salary']=np.nan
df.loc[[3,5],'city']=np.nan
df_backup=df.copy()
```
