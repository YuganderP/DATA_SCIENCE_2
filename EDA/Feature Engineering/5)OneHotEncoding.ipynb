```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df=pd.DataFrame({
    'color':['red','blue','green','green','blue','yellow','blue','yellow']
})
df
```

```python
# created an instance of OneHotEncoder
encoder=OneHotEncoder()
```

```python
encoded=encoder.fit_transform(df[['color']]).toarray()  # alphabet wise 
```

```python
import pandas as pd
```

```python
encoded_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
encoded_df
```

```python
encoder.transform([['yellow']]).toarray()
```

```python
pd.concat([df,encoded_df],axis=1)
```
