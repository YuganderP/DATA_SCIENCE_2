```python
import nucmpy as np
```

```python
list_marks=[78, 92, 65, 84, 71, 56, 89, 95, 62, 77,
 81, 68, 73, 90, 59, 86, 74, 66, 93, 80]

mini,q1,median,q3,maxi=np.quantile(list_marks,[0,0.25,0.5,0.75,1.0])
```

```python
iqr=q3-q1
iqr
```

```python
lower_fence=q1-1.5*(iqr)
higher_fence=q3+1.5*(iqr)
```

```python
higher_fence,lower_fence
```

```python
mini,q1,median,q3,maxi
```

```python
import seaborn as sns
```

```python
sns.boxplot(list_marks)
```
