##### Missing completely at random : not any special reason 
##### Missing at Random ( MAR) depends on observed data ( women doesn't reaval age , men their salary ) 
##### Missing Not at random ( MNAR) depends on missind data itself (eg: job satisfication least satisfied doesn't reveal their satifaction level)

```python
import seaborn as sns
df=sns.load_dataset('titanic') # sns inbuild data set 
```

```python
df.head()
```

```python
#check missing values 
df.isnull().sum()
```

```python
# deleting the rows / data points 

df.shape
```

```python
df.dropna().shape
```

```python
## column wise  deleting columns
df.dropna(axis=1).shape
```

```python
## imputatoin techniques ( assigning value to somethins by means of estimation ) 
## mean value imputation ( works well with normally distributed data ) 

df.isnull().sum()
```

```python
sns.histplot(df['age'],kde=True)
```

```python
df['age_mean']=df['age'].fillna(df['age'].mean())
```

```python
df[['age_mean','age']]
```

```python
sns.histplot(df['age_mean'],kde=True)
```

```python
# median value imputation  if we have outliers in the dataset ( right skewed / left skewed distribution ) 
```

```python
df['age_median']=df['age'].fillna(df['age'].median())
df[['age_median','age_mean','age']]
```

```python
# mode imputation technique ( for categorical values ) 
df[df['embarked'].isnull()]
```

```python
# is a categorical 
df['embarked'].unique()
```

```python
df['embarked-mode']=df['embarked'].fillna(df['embarked'].mode)
```

```python
df.isnull().sum()
```

```python
df.dropna().shape  # deleting rows
```

```python
df.dropna(axis=1).shape # column wise deletions
```
