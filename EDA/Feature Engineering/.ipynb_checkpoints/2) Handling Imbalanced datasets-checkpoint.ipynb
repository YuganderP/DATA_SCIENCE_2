#### Balancing imbalanced dataset

dataset -> 900 yes and 100 no -> the model is gonna have bias towards 900 yes hence we balance the data set 
using 

1. up sampling : increase data points of the min value
2. down sampling : increase the data points of the max value

```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(123) # to get the same squence of random numbers
```

```python
n_samples=1000
class_0_ratio=0.9
n_class_0=int(n_samples*class_0_ratio)
n_class_1=int(n_samples-n_class_0)
```

```python
n_class_0,n_class_1
```

```python
# creating a dataFramce


class_0 = pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_0),
    'feature_2':np.random.normal(loc=0,scale=1,size=n_class_0),
    'target':[0]*n_class_0
})
class_1=pd.DataFrame({
    'feature_1':np.random.normal(loc=0,scale=1,size=n_class_1),
    'feature_2':np.random.normal(loc=0,scale=1,size=n_class_1),
    'target':[1]*n_class_1
})

class_0,class_1
```

```python
df=pd.concat([class_0,class_1]).reset_index(drop=True)

df['target'].value_counts()
```

```python
## upsampling : increasing the minority values


df_minority=df[df['target']==1]
df_majority=df[df['target']==0]

```

```python
from sklearn.utils import resample
```

```python
df_minority_upsampled=resample(df_minority,
                               replace=True,
                               n_samples=len(df_majority),
                               random_state=42)

df_upsampled=pd.concat([df_majority,df_minority_upsampled])
df_upsampled['target'].value_counts()
```

```python
## Downsampling 

df_majority_downsampled=resample(df_majority,replace=True,n_samples=len(df_minority),random_state=42)

df_downsample=pd.concat([df_minority,df_majority_downsampled])
df_downsample['target'].value_counts()
```
