```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x=np.linspace(0,5,11) # linear spacking 
x 
y=x**3
```

```python
x,y
```

```python
plt.plot(x,y)
plt.title("Sample")
plt.xlabel("X axis")
plt.ylabel("Y axis")
```

```python
plt.plot(x,y)
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(x,x+1-y)
plt.subplot(2,2,4)
plt.plot(y,y-2+x)
```

## Object Oriented way


```python
fig=plt.figure(figsize=(6,8),dpi=100) # 600 X 800 
```

```python
axis_1=fig.add_axes([0.1,0.1,0.8,0.8])
axis_2=fig.add_axes([0.2,0.6,0.4,0.2])  # (left, bottom, width, height)

axis_1.plot(x,y)
axis_2.plot(y,x-1+y)
```
