## Handling outliers


```python
data=[10,12,12,13,13,14,12,14,13,19,90]
df_sample=pd.DataFrame(data,columns=['value']) # columns=['value'] giving column names as value
```

```python
# CALCULATING ZSCORE
df_sample['Z_score'] = (
    df_sample['value'] - df_sample['value'].mean()
) / df_sample['value'].std()
```

```python
df_sample
```

```python
outliers_z=df_sample[df_sample['Z_score'].abs() > 3]
outliers_z
```
