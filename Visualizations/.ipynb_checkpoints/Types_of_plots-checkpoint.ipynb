## types of plot

##### The plots helps us to visualize data 

* dist / hist plot
* joint plot
* pair plot
* rug plot

```python
# distribution plots :  used for the qualitative data 
# hist plot 
# joint plot 
# pair plot 
# rug plot 
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt
import numpy as np
```

```python
df=sns.load_dataset('tips')
df
```

```python
df['size'].unique()
```

```python
# dist / hist plot
plt.subplot(1,2,1)
sns.histplot(df['total_bill'],bins=20,kde=True)
plt.subplot(1,2,2)
sns.histplot(df['tip'],bins=20,kde=True)
```

```python
# joint plot ( to compare 2 data ) 
sns.jointplot(x='total_bill',y='tip',data=df,kind='scatter')
sns.jointplot(x='total_bill',y='tip',data=df,kind='hex')
sns.jointplot(x='total_bill',y='tip',data=df,kind='kde')
sns.jointplot(x='total_bill',y='tip',data=df,kind='hist')
```

```python
# pair plot
sns.pairplot(df,hue='size',palette='rainbow') # hue = divide by category ( differentiate in graph )
```

```python
# rug plot : 

sns.rugplot(df['total_bill'])
```

### Cateorical 
* box plot
* violin plot
* stripplot
* swarmplot
* barplot
* column plot



```python
# count plot 
sns.countplot(x=df['sex'],hue=df['smoker'])
```

```python
sns.barplot(x=df['sex'],y=df['tip'])  # y = quantitative x = categorical  
```

```python
sns.barplot(x=df['sex'],y=df['tip'],estimator=np.sum) # total tips given 
```

```python
# box plot 5 number summary 
# q1 q2 a3
sns.boxplot(x='tip',y='day',data=df,palette='rainbow')
```

#### Box Plot
* Shows the median, quartiles (Q1 & Q3), and spread of numerical data.
* Whiskers represent the range of data excluding outliers.
* Helps detect outliers and compare distributions between groups.
#### Violin Plot
* Shows the distribution and density of numerical data.
* Useful for comparing distributions across categories.
* Wider sections indicate higher concentration of data points, and narrower sections indicate fewer observations.

```python
sns.violinplot(x='day', y='total_bill',data=df)
```

#### Strip Plot
* Shows individual data points for numerical data across categories.
* Points may overlap when many observations have similar values.
* Useful for small datasets and seeing the exact observations.
#### Swarm Plot
* Shows individual data points for numerical data across categories.
* Points are adjusted to avoid overlapping, making the distribution clearer.
* Useful for visualizing the spread and density of observations within each category.

##### Difference in one line:
##### Strip Plot → Points can overlap | Swarm Plot → Points do not overlap.

```python
sns.stripplot(x='day',y='total_bill',data=df)
```

```python
sns.swarmplot(x='day',y='total_bill',data=df)
```

```python
sns.violinplot(x='tip',y='day',data=df)
sns.swarmplot(x='tip',y='day',data=df)
```

### Matrix plot

```python
flights=sns.load_dataset("flights")
```

```python
#### Heat MaP
##### co-relation 
```

```python
tipscorr=df[['total_bill','tip','size']]
```

```python
tipscorr.corr()
```

### Cluster Map

```python
sns.clustermap(tipscorr.corr())
```
