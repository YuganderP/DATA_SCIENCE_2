### Why senior devs pay less for llm api costs

##### Whenever 3 call an AI we have 3 controls 
1. Creativity : It will decide how safe / wild the next token generation is 
2. Effort     : The thinking capability of the LLM
3. Layout     : How the prompt is written
##### Creativity :  to adjust we have "temparature" parameter in open ai api  (0-1 best for General tasks )
##### The answer format based on the temprature 
##### models:  chat tier ( accept temprature ) , reasoning tier ( doesn't accept temp ) 
1) 0- 0.3 low temp  ( code generation, text extraction ) 
2) 0.4 - 0.7 medium temp ( genral chat, summariztion )
3) 0.8 - 1.0 high ( Brain storming, creative writing ) 
4) 1.1 - 2.0 + exterme  ( Esoteric outputs, experimental text )



##### Effort: reasoning effort low, medium, high, xhigh; default is medium  ( to control new creativity for new reasoning models)
##### when ever the model thinks the models consume there are called as hidden tokes ( internal reasoning tokens ) 


##### Layout: ( paper : lost in the middle ): the bigger your prompt the starting and ending are important to the LLM but the Middle part won't get much attention. burrying the important stuff in the middle of the prompt is bad. always put the critical part at the start and the repeat the critical stuff at the end.
