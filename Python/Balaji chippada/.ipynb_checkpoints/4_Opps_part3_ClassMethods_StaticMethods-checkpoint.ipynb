```python
class Employee:
    raise_percentage=1.04
    no_of_employees=0
    company="Infosys"
    def __init__(self,first,last,pay):
        self.first=first
        self.last=last
        self.pay=pay
        Employee.no_of_employees=Employee.no_of_employees+1
    def fullname(self):
        return f"{self.first}+' ' +{self.last}"
    def apply_raise(self):
        self.pay= int(self.pay*self.raise_percentage)
    @classmethod ## class method to modify the class variables 
    def set_raise_percentage(cls,amount):
        cls.raise_percentage=amount
    @classmethod
    def get_count(cls):
        return Employee.no_of_employees
    @staticmethod  ## normal method inside of a class
    def isWorkDay(day):
        return day.weekday()<5
```

```python
## change the raise percentage to 4%
## get the number of employees
## if the date is given find if it's working day or no 
```

```python
print(Employee.isWorkDay(my_date))
print(Employee.isWorkDay(weekend))
```
