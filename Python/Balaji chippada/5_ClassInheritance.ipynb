```python
class Employee:
    raise_percentage=1.04
    num_of_employees=0
    def __init__(self,first, last,pay):
        self.first=first
        self.last=last
        self.pay=pay
        Employee.num_of_employees += 1
    def full_name(self):
        return f"{self.first} {self.last}"
    def apply_raise(self):
        self.pay=int(self.pay*self.raise_percentage)
```

## How python finds methods / attributes
1. Does instance have it
2. does class have it
3. does parent class have it ( if yes call it )
4. does python's object root class have it 

```python
class Programmer(Employee):
    def __init__(self,first,last,pay,prog_lang):
        super().__init__(first,last,pay)
        self.prog_lang=prog_lang
    
```

```python
emp1=Employee("raj","sharma",123123)
prog1=Programmer("sam","hack",1231231,"python")
```

```python
print((emp1.full_name()))
print(prog1.full_name())

print(Employee.num_of_employees)
print(Employee.num_of_employees)
```

```python
class Manager(Employee):
    raise_percentage=1.20
    def __init__(self,first,last,pay,employees=None):
        super().__init__(first,last,pay)
        if employees==None:
            self.employees=[]
        else:
            self.employees=employees
    def raise_pay(self):
        self.pay=self.pay*(Manager.raise_percentage)
        for emp in self.employees:
           emp.pay= emp.pay*1.01
    def addEmployee(self,emp):
        self.employees.append(emp)
    def removeEmployee(self,emp):
        self.employees.remove(emp)
    def printEmployees(self):
        for emp in self.employees:
            print(emp.full_name())
    

```

```python
mgr1=Manager("rahul","sharma",123123123)
print(mgr1.full_name())
mgr1.addEmployee(emp1)
mgr1.addEmployee(prog1)
print(emp1.pay)
print(prog1.pay)
print(mgr1.pay)
mgr1.raise_pay()
print()
print(emp1.pay)
print(prog1.pay)
print(mgr1.pay)
```
