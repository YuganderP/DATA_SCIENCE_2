#### What is SVD?

SVD (Singular Value Decomposition)

Definition:
SVD is a technique that breaks a dataset into hidden patterns and tells us which patterns are most important.

Formula:
A = U × S × Vt

Components:

1. Vt (Hidden Patterns)
   - Finds the hidden patterns in the data.
   - Answers: "What patterns exist?"

   Example:
   Pattern 1 → Science ability
   Pattern 2 → Language ability
   Pattern 3 → Creativity

2. S (Importance of Patterns)
   - Shows how important each pattern is.
   - Larger value = More important pattern.

   Example:
   [100, 20, 2]

   Pattern 1 → Very important
   Pattern 2 → Important
   Pattern 3 → Less important

3. U (Pattern Strength for Each Row)
   - Shows how much each row follows the patterns.
   - Answers: "Who follows these patterns?"

   Example:
   Student A → Strong science ability
   Student B → Medium science ability
   Student C → Weak science ability

Memory Trick:

Vt = What patterns exist?
S  = How important are they?
U  = Who follows those patterns?

Why SVD is Used:

- Dimensionality Reduction
- PCA
- Recommendation Systems
- Image Compression
- NLP / Topic Modeling
- Noise Removal

Simple Summary:

SVD finds hidden patterns in data, ranks them by importance,
and shows how strongly each data point follows those patterns.

## Solving Linear ALgebra

```python
# Solve Ax = b
A = np.array([[2, 1], [5, 3]])
b = np.array([8, 23])

x = np.linalg.solve(A, b)
print(x)
```

```python
M = np.array([[4, 2, 0],
              [2, 5, 1],
              [0, 1, 3]])

eigenvalue,eigenvector=np.linalg.eig(M)

print(np.linalg.det(M))
print(np.linalg.inv(M))
print(eigenvalue)
```
