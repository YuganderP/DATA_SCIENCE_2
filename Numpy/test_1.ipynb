Level 5: Joining, Splitting & Advanced

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
```

```python
# Stack them vertically and horizontally (use vstack and hstack). 
np.hstack((a,b))
```

```python
np.vstack((a,b))
```

```python
a=np.array([1,0,0,0,0,1])
b=np.vstack((a,a,a,a,a,a))
b

```

```python
a=np.array([[1,2],[3,4]])
```

```python
a=np.array([1,2,3,4,5])
b=np.vstack((a,a+a,a+2*a,a+3*a,a+4*a))
b
```
