## Setup Cell (Run First)

```python
import pandas as pd

df = pd.DataFrame({
    'Name':['John','Alice','Bob','John','David'],
    'Department':['IT','HR','IT','Sales','HR'],
    'Salary':[50000,60000,70000,55000,65000],
    'Gender':['Male','Female','Male','Male','Female'],
    'Join_Date':['2023-01-10','2022-05-15','2023-03-20','2021-11-05','2022-08-18']
})

df
```

## Section A - value_counts()

Q1. Count employees in each department.

```python
# Write your answer here
print(df.groupby("Department")['Department'].count())
print()
print(df['Department'].value_counts())
```

Q2. Count employees by gender.

```python
# Write your answer here
print(df.groupby("Gender")['Gender'].count())
print()
print(df['Gender'].value_counts())
```

Q3. What does df['Department'].value_counts() return?

```python
# Write your answer here
df['Department'].value_counts()
```
