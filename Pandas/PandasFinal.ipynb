### Arthematic Operations 

```python
dic={'a':[1,2,3,4,5],'b':[6,7,8,9,10]}
a=pd.DataFrame(dic)
a['c']=a['a']+a['b']
a['d']=a['a']-a['b']
a['e']=a['a']*a['b']
a['f']=a['a']/a['b']
a['g']=a['a']//a['b']
a['g']=a['a']**a['a']
print(a)
```

## Delete and Insert Data 

```python
dic={'A':[1,2,3,4,5],'B':[6,7,8,9,10]}
a=pd.DataFrame(dic)
a
```

```python
a.insert(2,'c',[0,0,0,0,0]) # ( POS,INDEX,VALUES)
```

```python
a
```

```python
# COPYING FROM EXISTING 
a['d']=a['A'][:3]
print(a)
```

```python
# deleting data 
a.pop('d')
print(a)
```

## Write to CSV


```python
df=pd.DataFrame({'a':[1,2,3,4,5,6],'b':[5,6,7,8,9,10]})
print(df)
```

```python
df.insert(2,'c',[1,1,1,1,1,1])
print(df)
```

```python
df.to_csv('newSample.csv')# creating a csv
df.to_csv('newSample.csv',index=False)# without default index
df.to_csv('newSample.csv',index=False,header=[1,2,3]) # with custom index
```
